```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Dataset teks ulasan sederhana untuk analisis sentimen
corpus = [
    "Pelayanan sangat memuaskan dan ramah",
    "Barang rusak dan pengiriman sangat lambat",
    "Saya sangat suka produk ini, luar biasa",
    "Kualitas buruk, buang-buang uang saja",
    "Kurir ramah dan produk berfungsi dengan baik"
]
labels = [1, 0, 1, 0, 1]  # 1: Positif, 0: Negatif

# Ekstraksi Fitur Teks menggunakan TF-IDF Vectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)

print("--- Representasi Fitur TF-IDF ---")
print(X.toarray())

# Melatih model klasifikasi teks sederhana
model_sentimen = LogisticRegression()
model_sentimen.fit(X, labels)

# Melakukan prediksi pada kalimat baru
kalimat_baru = ['Pelayanan sangat memuaskan dan ramah']
X_new = vectorizer.transform(kalimat_baru)
pred = model_sentimen.predict(X_new)

print('\nHasil Prediksi Kalimat Baru:', 'Positif' if pred[0] == 1 else 'Negatif')
```

### Kesimpulan
- **Apa yang dipelajari:** Mempelajari pengantar transisi dari *Machine Learning* klasik ke *Deep Learning* (*Artificial Neural Network*) serta teknik pemrosesan teks (*Text Vectorization* menggunakan TF-IDF) untuk melakukan tugas analisis sentimen.
- **Temuan utama:** Representasi matriks numerik berbasis TF-IDF berhasil dimanfaatkan oleh model klasifikasi untuk mendeteksi polaritas teks ulasan (*Positif* atau *Negatif*) secara otomatis.
- **Keterbatasan / Pertanyaan:** Pendekatan TF-IDF mengabaikan konteks struktur susunan kata dan urutan kalimat, sehingga untuk teks yang kompleks memerlukan arsitektur *Deep Learning* tingkat lanjut seperti *Embeddings* atau *Transformer*.
