```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Membuat dataset transaksi ritel sederhana (Market Basket Analysis)
data = {
    'Roti': [1, 1, 0, 1, 1, 0],
    'Susu': [1, 1, 1, 0, 1, 1],
    'Kopi': [0, 1, 1, 1, 0, 1],
    'Gula': [0, 0, 1, 1, 0, 1],
    'Teh': [1, 0, 0, 1, 1, 0]
}
df = pd.DataFrame(data)

print("--- DataFrame Transaksi ---")
display(df)

# Mencari Frequent Itemsets menggunakan Algoritma Apriori
frequent_itemsets = apriori(df, min_support=0.3, use_colnames=True)
print("\n--- Frequent Itemsets ---")
display(frequent_itemsets)

# Membuat Association Rules berdasarkan Confidence
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)
print("\n--- Association Rules ---")
display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

### Kesimpulan
- **Apa yang dipelajari:** Memahami konsep *Association Rule Mining* menggunakan Algoritma Apriori (*Market Basket Analysis*) untuk mencari pola kebiasaan pembelian barang dengan parameter *Support*, *Confidence*, dan *Lift*, serta pengantar sistem rekomendasi dasar.
- **Temuan utama:** Nilai *Lift* > 1 menunjukkan adanya hubungan asosiasi yang kuat dan bermakna antar item produk yang sering dibeli secara bersamaan oleh konsumen.
- **Keterbatasan / Pertanyaan:** Proses pencarian aturan asosiasi dapat menjadi lambat (*combinatorial explosion*) apabila jumlah variasi produk dan transaksi data ritel berukuran sangat besar.
